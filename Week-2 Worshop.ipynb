```python
%sql
-- E1: average days from purchase to delivery
SELECT ROUND(AVG(datediff(order_delivered_customer_date, order_purchase_timestamp)), 2) AS avg_delivery_days
FROM orders_2025
WHERE order_status = 'delivered'
AND order_delivered_customer_date IS NOT NULL;
```

```python
%sql
-- E2: Which day in 2025 had the most orders placed
SELECT to_date(order_purchase_timestamp) AS order_day,
       COUNT(*) AS total_orders
FROM orders_2025
WHERE year(order_purchase_timestamp) = 2025
GROUP BY order_day
ORDER BY total_orders DESC
LIMIT 1;
```

```python
%sql
-- E3: How many orders arrived after their estimated delivery date?
SELECT COUNT(*) AS late_orders
FROM orders_2025
WHERE order_status = 'delivered'
  AND to_date(order_delivered_customer_date) > to_date(order_estimated_delivery_date);
```

```python
%sql
-- E4: Which month has the highest cancellation rate and why might that be?
SELECT date_format(order_purchase_timestamp, 'yyyy-MM') AS order_month,
       COUNT(*) AS total_orders,
       SUM(CASE WHEN order_status = 'canceled' THEN 1 ELSE 0 END) AS canceled_orders,
       ROUND(100.0 * SUM(CASE WHEN order_status = 'canceled' THEN 1 ELSE 0 END) / COUNT(*), 2) AS cancel_rate_pct
FROM orders_2025
GROUP BY order_month
ORDER BY cancel_rate_pct DESC;
```
